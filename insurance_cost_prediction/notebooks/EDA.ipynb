### Import necessary libraries 

```python
import pandas as pd 
from pathlib import Path 
import numpy as np 
from dotenv import load_dotenv
import json
```

### Environment setup and configuration loading

```python
# Load the .env variables and define the base working directory 
load_dotenv()
BASE_DIR = Path.cwd().parent

# Define the configuration path and read configuration variables from the json file
CONFIG_PATH = BASE_DIR/"config"/"insurance_config.json"
with open(CONFIG_PATH) as m:
    config = json.load(m)

# Define the path to the raw data 
RAW_PATH = BASE_DIR/config["raw_path"]
```

### Loading the raw dataset

```python
dataset = pd.read_csv(RAW_PATH)
dataset.sample(5)
```
