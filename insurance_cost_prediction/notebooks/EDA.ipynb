##### Import required libraries

```python
import pandas as pd 
import numpy as np 
from pathlib import Path
from dotenv import load_dotenv
import json 
```

##### Load environment variables, configuration settings, and define project directory paths


```python
##  Load the .env variables and define the working base directory
load_dotenv()
BASE_DIR = Path.cwd().parent

##  Define the path to config and load the configuration variables from json
CONFIG_PATH = BASE_DIR/"config"/"insurance_config.json"
with open(CONFIG_PATH) as m:
    config = json.load(m)

## Define the path to the raw dataset 
RAW_PATH = BASE_DIR/config["raw_path"]
```

##### Load the raw dataset

```python
dataset = pd.read_csv(RAW_PATH)
dataset.sample(5)
```
