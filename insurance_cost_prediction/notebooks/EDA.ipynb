#### Import required libraries

```python
import pandas as pd 
import numpy as np 
import json 
from pathlib import Path
```

#### Identify the path to config and load it's variables from json

```python
CONFIG_PATH = BASE_DIR/"config"/"insurance_config.json"
with open(CONFIG_PATH) as m: 
    config = json.load(m)
```
